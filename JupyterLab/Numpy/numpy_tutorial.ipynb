```python
import numpy as np
import time
```

```python
#execution performance
size = 10_000_000

#Python list
py_list = list(range(size))
start = time.time()
sq_list = [x**2 for x in py_list]
end = time.time()

print(f"Python list time = {end-start} seconds")
#numpy arrays
np_arr = np.array(py_list)
start = time.time()
#Vectorization

sq_array = np_arr ** 2
end = time.time()
print(f"Numpy array time = {end-start} seconds")
```

```python
#Memory

import sys
print(f"Python size list = {sys.getsizeof(py_list) * len(py_list)} bytes")

print(f"numpy array size = {np_arr.nbytes} bytes")
```
